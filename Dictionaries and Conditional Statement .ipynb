### Analyze the output and explain the same.
### &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;   ✅Duplicates are automatically removed.
### &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;   ✅It only keeps unique items.

### &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;   ✅Input list has duplicates
### &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;   ✅Set output has no duplicates

### B.Attempt to access my_set[4]. If the code throws an error, provide an explanation.
#### &nbsp;&nbsp;&nbsp;syn : print(my_set[4])
&nbsp;
&nbsp;
## 🔴<u> Error(TypeError: 'set' object is not subscriptable)</u>
#### <u>Explanation</u>
#### &nbsp; 🟡A set in Python does NOT support indexing.
#### &nbsp; 🟡 Sets are unordered collections,meaning the elements do not have fixed positions like index 0, 1, 2, 3, 
#### &nbsp; 🟡 When you try to access my_set[4], Python looks for the element at index 4 — but sets do not have indexes.
&nbsp;
&nbsp;

## <u>That means:</u>
#### &nbsp; 🟡There is no fixed position like index 0, 1, 2, 3, 4…
#### &nbsp; 🟡So you cannot access elements using square brackets [].
&nbsp;
&nbsp;
#### &nbsp; 🟢If you want to see elements in a set
#### &nbsp; 🟢You must use a loop, not indexing:
## ✅<u>syntax:</u>
# for item in my_set:
# print(item)


```python
#b. Attempt to access my_set[4]. If the code throws an error, provide an explanation.
print(my_set[4])
```

```python
for item in my_set:
    print(item)
```

```python
#c. Create two sets:

set1={1, 3, 5, 7, 9}
set2={2, 3, 5, 8, 10}
print(type(set1))
```

### union() is a set operation that combines elements from two or more sets.

### <u> Key Point</u>
### &nbsp; ✅Duplicates are removed automatically
### &nbsp; ✅The result contains all unique elements from all sets involved

```python
#d. Compute and print the union and intersection of set1 and set2.
print('set 1 :',set1)
print('set 2 :',set2)
print('\nUnion set :',set1.union(set2))
```

## Intersection() is a set operation that returns only the elements that are common between two sets.
#### 🔵It keeps only the values that appear in BOTH sets
#### 🔵All duplicates are removed automatically
#### 🔵The result is always a set of unique common elements

```python
#intersection
print('set 1 :',set1)
print('set 2 :',set2)
print('\nIntersection set :',set1.intersection(set2))
```

#### 🔹IF Statement - Runs a block of code only if the condition is true.

#### 🔹ELSE Statement - Runs when none of the above conditions are true.

### Performance Category Program:

#### a. Above Average: Score greater than 7
#### b. Average: Score between 4 and 7 (both inclusive)
#### c. Below Average: Score less than 4

```python
#Prompt the user for input. Score range should be from 0 to 10 (both inclusive).
print('\nPerformance category output')
score_range=int(input('Enter your score(0 to 10) :'))
if score_range>7:
    print('Above Average: Excellent performance! Keep up the great work')
elif score_range>=4 and score_range<=7:
    print("Average : Good effort! Keep practicing, there's room for improvement.")
elif score_range<4:
    print("Below Average :'Need to improve your performance; consistent practice will lead to better results'.")
else:
    print("Invalid score! Please enter a value between 0 and 10.")
    
```
